## Question12-Left rotate an array by one position

```python
arr=[10,20,30,40,50]

first=arr[0]

for i in range(len(arr)-1):
    arr[i]=arr[i+1]
arr[-1]=first
    
print(arr)
```

## Question14-Remove duplicate elements from sorted array

```python
arr = [1, 1, 2, 2, 3, 4, 4, 5]

j = 0

for i in range(1, len(arr)):
    if arr[i] != arr[j]:
        j += 1
        arr[j] = arr[i]

print(arr[:j + 1])
```

## Question15-Find the Missing Number in an Array (1 to N)

```python
arr = [1, 2, 4, 5]
N = 5

total_sum = N * (N + 1) // 2
array_sum = sum(arr)

missing = total_sum - array_sum

print("Missing Number:", missing)
```
