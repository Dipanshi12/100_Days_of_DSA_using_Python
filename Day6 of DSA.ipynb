## Quetion26-Rotate an Array by K Positions

```python
arr = [1, 2, 3, 4, 5]
k = 2

n = len(arr)
k = k % n

result = arr[n-k:] + arr[:n-k]

print(result)
```

## Question27-Find the Sum of All Elements in an Array

```python
arr = [10, 20, 30, 40, 50]

total = 0

for i in arr:
    total = total + i

print(total)
```

## Question28-Find the Average of Array Elements

```python
arr = [10, 20, 30, 40, 50]

total = 0

for i in arr:
    total = total + i

average = total / len(arr)

print(average)
```

## Question29-Find the Maximum Difference Between Two Elements

```python
arr = [7, 1, 5, 3, 6, 4]

min_element = arr[0]
max_difference = 0

for i in range(1, len(arr)):
    difference = arr[i] - min_element

    if difference > max_difference:
        max_difference = difference

    if arr[i] < min_element:
        min_element = arr[i]

print(max_difference)
```

## Question30-Search an Element Using Binary Search

```python
arr = [10, 20, 30, 40, 50]
target = 30

start = 0
end = len(arr) - 1

while start <= end:
    mid = (start + end) // 2

    if arr[mid] == target:
        print("Element found at index:", mid)
        break

    elif target < arr[mid]:
        end = mid - 1

    else:
        start = mid + 1

else:
    print("Element not found")
```
