## Question1-Find largest element in array

```python
arr = [10,25,8,45,12]
largest=arr[0]
for i in arr:
    if i > largest:
        largest = i
print("Largest Element is:",largest)
        
```

## Question2-Reverse an array

```python
arr =[1,2,3,4,5]
start=0
end=len(arr)-1
while start< end:
    arr[start],arr[end]=arr[end],arr[start]
    start+=1
    end-=1
print(arr)
    
```

## Question4-Count Even and Odd Numbers in an Array

```python
arr = [10, 15, 22, 7, 8, 13]
Even = 0
Odd = 0
for i in arr:
    if i % 2==0:
        Even +=1
    else:
        Odd+=1
print("Count of Even:",Even)
print("Count of Odd:",Odd)
```
