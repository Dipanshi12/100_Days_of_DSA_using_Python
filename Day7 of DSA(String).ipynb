## Question2-Reverse string

```python
a = "hello"          # Without slicing
reverse = ""
for char in a:
    reverse=char+reverse
print(reverse)
```

```python
a = "hello"      # Slicing
print(a[::-1])
```

## Question3-Palindrome

```python
a = "madam"          # slicing
if a == a[::-1]:
    print("True")
else:
    print("False")
```

```python
s = "madam"

start = 0
end = len(s) - 1

is_palindrome = True

while start < end:

    if s[start] != s[end]:
        is_palindrome = False
        break

    start += 1
    end -= 1

print(is_palindrome)
```

## Question4-Count vowels and consonants

```python
s = "programming"

vowels = 0
consonants = 0

for char in s:

    if char.lower() in "aeiou":
        vowels += 1

    elif char.isalpha():
        consonants += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
```

## Question5-Find frequency of each character

```python
s = "banana"

freq = {}

for char in s:
    if char in freq:
        freq[char] += 1
    else:
        freq[char] = 1

print(freq)
```
