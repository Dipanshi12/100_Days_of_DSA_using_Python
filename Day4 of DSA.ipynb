## Question16-Find frequency of the element in array

```python
arr = [10, 20, 30, 20, 40, 20, 50]
target = 20
count=0
for i in range(len(arr)):
    if arr[i]==target:
        count+=1

print(f"Frequeny of target element is :{count}")
```

## Question18-Union of arrays

```python
# Method(1)
arr1 = [1, 2, 3, 4]
arr2 = [3, 4, 5, 6]
union=list(set(arr1+arr2))
print("Union of array is:",union)
```

```python
# Method(2)
arr1 = [1, 2, 3, 4]
arr2 = [3, 4, 5, 6]
union=[]
for i in arr1:
    if i not in union:
        union.append(i)
for i in arr2:
    if i not in union:
        union.append(i)
print("Union of two arrays is:",union)
```

## Question19-Intersection of arrays

```python
# Method(1)
arr1 = [1, 2, 3, 4]
arr2 = [3, 4, 5, 6]
intersection=list(set(arr1) & set(arr2))
print("Intersection is:",intersection)
```

```python
# Method(2)
arr1 = [1, 2, 3, 4]
arr2 = [3, 4, 5, 6]
intersection=[]
for i in arr1:
    if i in arr2 and i not in intersection:
        intersection.append(i)
print("Intersection is:",intersection)        
```
