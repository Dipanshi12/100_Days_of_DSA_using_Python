## Question22-Kadane's Algorithm (Maximum Subarray Sum)

```python
arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

current_sum = arr[0]
maximum_sum = arr[0]
for i in range(1,len(arr)):
    current_sum=max(arr[i] , current_sum+arr[i])
    if current_sum>maximum_sum:
        maximum_sum=current_sum
print("Maximum subarray sum is:",maximum_sum)
```

## Question23-Majority Element in an Array

```python
arr = [2, 2, 1, 1, 1, 2, 2]

count = {}

n = len(arr)

for num in arr:
    if num in count:
        count[num] += 1
    else:
        count[num] = 1

    if count[num] > n//2:
        print("Majority element is", num)
        break
```

## Question24-Rearrange Array Elements by Sign

```python
arr = [3, 1, -2, -5, 2, -4]

positive = []
negative = []

for num in arr:
    if num > 0:
        positive.append(num)
    else:
        negative.append(num)

result = []

for i in range(len(positive)):
    result.append(positive[i])
    result.append(negative[i])

print(result)
```

## Question25-Best Time to Buy and Sell Stock

```python
# Brute force approach
prices = [7, 1, 5, 3, 6, 4]

max_profit = 0

for i in range(len(prices)):
    for j in range(i + 1, len(prices)):
        profit = prices[j] - prices[i]

        if profit > max_profit:
            max_profit = profit

print("Maximum profit is", max_profit)
```
